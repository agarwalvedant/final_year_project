```python
from docx import Document
from transformers import pipeline

# Initialize the paraphrasing pipeline
paraphraser = pipeline("text2text-generation", model="t5-small", max_length=200)

def humanize_text(text):
    """Paraphrase the given text."""
    try:
        result = paraphraser(f"paraphrase: {text}", num_return_sequences=1)[0]['generated_text']
        return result
    except Exception as e:
        print(f"Error during paraphrasing: {e}")
        return text

def humanize_docx(input_path, output_path):
    # Load the document
    doc = Document(input_path)

    # Process each paragraph
    for para in doc.paragraphs:
        if para.text.strip():  # Only process non-empty paragraphs
            para.text = humanize_text(para.text)

    # Save the humanized document
    doc.save(output_path)
    print(f"Humanized document saved as: {output_path}")

# Example usage
input_path = "input.docx"
output_path = "humanized_output.docx"
humanize_docx(input_path, output_path)
```
